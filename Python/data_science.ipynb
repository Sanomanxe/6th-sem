Bar graph

```python
import matplotlib.pyplot as plt
subjects = ["Math","Science","English","computer"]
marks=[87,67,90,34]

plt.figure(figsize=(10,5))
plt.bar(subjects,marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Student marks comparision")
```

```python
#check if values are evenly distributed 
# if all bars are similar or tends to be -----> balanced
# if one is very high or low ------> outliers
```

```python
#when not use bar graph
# data is continuous over time ----> use line chart
# for distribution ----> use histogram
# want relationship ---> use scatter plot
```

Line graph

```python
exams=[1,2,3,4,5]
marks=[60,65,79,85,44]
plt.figure(figsize=(11,7))

plt.plot(exams,marks,marker='o')

plt.xlabel("Exam nbr")
plt.ylabel("marks")

plt.title("Marks trend over exams")

plt.show()
```

```python
exams=[1,2,3,4,5]
marks=[60,65,79,85,44]
plt.figure(figsize=(11,7))

plt.plot(exams,marks,marker='^')

plt.xlabel("Exam nbr")
plt.ylabel("marks")

plt.title("Marks trend over exams")

plt.show()
```

```python
# used to show how something changes over time or over continuous range or count
# it connects data points with lines, making patterns and trend easy to see
# line graph ==== trend (change)  over time

# from line graph what we can see: 
# trend (increase/decrease)
# fluctuation
# pattern over time
# rate of change how fast values increase or decrease

#pattern ./ cycle :
# repeated behaviour(eg : temperature rising or falling every afternoon)
#  best for time series data
# helps to detect anamolies (sudden spikes / drops)


```

```python
days=[1,2,3,4,5]

apple=[150,155,160,158,165]
google=[140,142,145,150,155]

plt.plot(days,apple,marker='o',label="Apple",linestyle='--')
plt.plot(days,google,marker='^',label="google")

plt.title("Days")
plt.ylabel("Stock price")
plt.title("Stock comparision : apple vs google")

plt.legend() #shows name on graph
plt.grid()
plt.show()
```

```python
#trend comparision
# apple rising faster than google
# if one line is always above ---> better performer
# 

# crossover:
# when two lines intersect

#volatity risk
# how much the line fluctuates
# more zig-zag ---> more unstable 

# divergence/convergence 
#lines moving apart --> divergence
#lines coming together ---> convergence
```

Scatter plot

```python
volume = [100,200,300,400,500,600,700]
price_change = [1,2,1.5,3,2.5,3.5,4]

volume2=[100,200,300,400,500,600,700]
price_change2=[0.5,1,1.2,1.8,2,2.2,2.5]

z=[260,490,550,450,530,610,730]

plt.figure(figsize=(5,5))

plt.scatter(volume,price_change,label="Stock A")
plt.scatter(volume2,price_change2,label="Stock B")

plt.xlabel("Trading volume")
plt.ylabel("Price change")

plt.title("Scatter plot : volume vs price change ")

plt.legend()
plt.grid()
plt.show()
```

```python
#scatter plot
# a scatter plot shows relationship between two variables
# each point == one observation
# x-axis = one features
# y-axis = another features

# it shows how the things are related
```

```python
# upward patterns ---> positive correlation
# downward patterns --> negative correlation
# random patterns/dots --> no relation
# far point --> outliers
# groups ---> clusters


# it is not about time. it is about relationship between variables
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[2,4,5,4,6,7,8]

plt.scatter(x,y,color="green")
plt.title("Positive correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

#as x increases -----> y increases
# eg : as temperature increases --> ice-cream sale increases
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[10,8,6,4,5,2,1]

plt.scatter(x,y,color="green")
plt.title("Negative correlation")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

#speed increases ---> time decreases
```

```python
import random
x=[1,2,3,4,5,6,7]
y=[random.randint(1,10) for i in x]
plt.scatter(x,y,color="green")
plt.title("NO correlation (random)")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()

#no pattern at all(random)
#show size vs IQ
```

Box Plot

```python
Science=[40,45,50,55,60,65,70,75,90]
math=[43,47,52,58,64,67,72,75]
plt.figure(figsize=(6,4))
plt.boxplot(Science)
plt.title("Box plot")
plt.show()
```

```python
#q1 = 25th percentile
#median=50th percentile
#q3 - 75th percentile

#box size = spread
# size of the box tells how spread out the middle (50%)
# small box = values are close together
# large box = values are more spread out

# whiskers --> normal range
# lines extending from the box is called whiskers
# whiskers : shows range of data within the non-outliers limits

# outliers ---> unusual values
# values that are unusually far from the rest

# if the median is 57.4 :
#   about half of the students scored below 57.4 and about half scored above 57.4
```
