```python
import pandas as pd
import numpy as np
df = pd.read_csv("marksheet.csv")
```

```python
import matplotlib.pyplot as plt

subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects]      # First student

plt.figure(figsize=(6,4))
plt.bar(subjects, marks)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks of First Student")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 
plt.figure(figsize=(11,7))

plt.plot(subjects,marks,marker='o')

plt.xlabel("Subject")
plt.ylabel("marks")

plt.title("Marks trend over subjects")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 
plt.figure(figsize=(11,7))

plt.plot(subjects,marks,marker='^')

plt.xlabel("Subjects")
plt.ylabel("marks")

plt.title("Marks trend over subjects")

plt.show()
```

```python
subjects = ["Science", "English", "History", "Maths"]
marks = df.loc[0, subjects] 


plt.figure(figsize=(5,5))

plt.scatter(subjects,marks,label=" ")

plt.xlabel("Subject")
plt.ylabel("Marks")

plt.title("Scatter plot : Subject vs marks ")

plt.legend()
# plt.grid()
plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.boxplot([df["Science"]])

plt.xlabel("Science")
plt.ylabel("Marks")
plt.title("Box Plot")

plt.show()
```
